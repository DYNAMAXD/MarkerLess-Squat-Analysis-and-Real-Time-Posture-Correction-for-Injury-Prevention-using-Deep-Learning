## 1. Pick a run

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display, Markdown

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
runs = sorted((ROOT / "results/runs").glob("*"))
print("available runs:"); [print("  ", r.name) for r in runs]
RUN = runs[-1]                      # <- or: RUN = ROOT / "results/runs/<folder name>"
print("\nusing:", RUN.name)
```

## 2. Model comparison table

```python
pd.read_csv(RUN / "tables/model_comparison.csv")
```
